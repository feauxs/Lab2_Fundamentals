```python
# TASK 1
# Setup (reuse from Cell 1)
ENGINEER = "Campos"
STUDENT_ID = "TUPM-26-1247"
SEED_NUM = int(STUDENT_ID[-1])

# Requirement 1: Generate sensor data using surname and SEED_NUM 
SENSOR_COUNT = SEED_NUM          # fixed number of readings, tied to the seed
LOW_LIMIT = 5
HIGH_LIMIT = (SEED_NUM * 10) + 20

generated_readings = []
for i in range(SENSOR_COUNT):
    char = ENGINEER[i % len(ENGINEER)]
    value = (ord(char) + (SEED_NUM * i)) % 100
    generated_readings.append(value)

raw_data = [str(v) for v in generated_readings]
raw_data[2] = "ERR"              # deliberately invalid entry to test validation

# Requirement 2 & 3: Process fixed readings, handle invalid input, classify 
valid_readings = []
invalid_count = 0
classification_summary = {"LOW": 0, "NORMAL": 0, "HIGH": 0}
execution_log = [f"OPERATING RANGE: LOW < {LOW_LIMIT} | NORMAL {LOW_LIMIT}-{HIGH_LIMIT} | HIGH > {HIGH_LIMIT}"]

for i, entry in enumerate(raw_data):
    try:
        reading = int(entry)
        valid_readings.append(reading)
        if reading < LOW_LIMIT:
            category = "LOW"
        elif reading > HIGH_LIMIT:
            category = "HIGH"
        else:
            category = "NORMAL"
        classification_summary[category] += 1
        execution_log.append(f"Reading {i+1}: {reading} -> {category}")
    except ValueError:
        invalid_count += 1
        execution_log.append(f"Reading {i+1}: '{entry}' -> INVALID (non-numeric)")

# Requirement 4: Display summary 
print("Generated Sensor Data:", raw_data)
print("Valid/Invalid Results:", f"{len(valid_readings)} valid {valid_readings} | {invalid_count} invalid")
print("Classification Results:", classification_summary)
print("Execution Log:")
for line in execution_log:
    print("  ", line)
print("Final Output:", f"Total Readings: {len(raw_data)} | Valid: {len(valid_readings)} | Invalid: {invalid_count}")
```

```python
# TASK 2
# Setup (reuse from Cell 1) 
ENGINEER = "Campos"
STUDENT_ID = "TUPM-26-1247"
SEED_NUM = int(STUDENT_ID[-1])

# Requirement 1: Generate the signal using surname and SEED_NUM 
generated_signal = f"  {ENGINEER} Seed-{SEED_NUM} Ch@nnel!  "

# Requirement 2: Process and normalize the generated signal 
processed_signal = generated_signal.strip()

# Requirement 3: Analyze the signal character by character 
def classify(ch):
    if ch.isalpha():
        return "LETTER"
    elif ch.isdigit():
        return "DIGIT"
    elif ch.isspace():
        return "SPACE"
    else:
        return "OTHER"

letters = digits = spaces = others = 0
execution_log = []
current_kind = None
current_chunk = ""

for ch in processed_signal:
    kind = classify(ch)
    if kind == "LETTER": letters += 1
    elif kind == "DIGIT": digits += 1
    elif kind == "SPACE": spaces += 1
    else: others += 1

    # group consecutive same-type characters into one log line
    if kind == current_kind:
        current_chunk += ch
    else:
        if current_kind is not None:
            execution_log.append(f"'{current_chunk}' -> {current_kind}")
        current_kind = kind
        current_chunk = ch
execution_log.append(f"'{current_chunk}' -> {current_kind}")   # flush last chunk

character_analysis = {"Letters": letters, "Digits": digits, "Spaces": spaces, "Other Symbols": others}

# Requirement 4: Display the analysis and final classification 
type_count = sum(1 for count in [letters, digits, others] if count > 0)

if type_count >= 3:
    signal_class = "MIXED SIGNAL (Alphanumeric + Symbols)"
elif letters > 0 and digits > 0:
    signal_class = "ALPHANUMERIC SIGNAL"
elif letters > 0:
    signal_class = "TEXT-ONLY SIGNAL"
else:
    signal_class = "NUMERIC-ONLY SIGNAL"

print("Generated Signal:", repr(generated_signal))
print("Processed Signal:", repr(processed_signal))
print("Character Analysis:", character_analysis)
print("Execution Log:", " | ".join(execution_log))
print("Signal Classification:", signal_class)
print("Final Output:", f"'{processed_signal}' classified as {signal_class}")
```

```python
# TASK 3
# Setup (reuse from Cell 1) 
ENGINEER = "Campos"
STUDENT_ID = "TUPM-26-1247"
SEED_NUM = int(STUDENT_ID[-1])

# Requirement 1: Generate the access credential and attempt limit 
generated_password = f"{ENGINEER.upper()}{SEED_NUM}"
attempt_limit = (SEED_NUM % 5) + 3   # unique attempt limit derived from the seed

# Requirement 2: Manage the authentication process 
attempts_made = 0
access_granted = False
execution_log = []

while attempts_made < attempt_limit:
    attempt = input("Enter Password: ")
    attempts_made += 1
    if attempt == generated_password:
        access_granted = True
        execution_log.append(f"Attempt {attempts_made}: '{attempt}' -> ACCESS GRANTED")
        print("ACCESS GRANTED.")
        break
    else:
        remaining = attempt_limit - attempts_made
        execution_log.append(f"Attempt {attempts_made}: '{attempt}' -> DENIED ({remaining} remaining)")
        print(f"Access Denied. Attempts remaining: {remaining}")

# Requirement 3: Determine the final state of the system 
final_state = "UNLOCKED" if access_granted else "LOCKED (Max attempts exceeded)"

# Display summary 
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)
print("Attempts Made:", attempts_made)
print("Access Result:", "GRANTED" if access_granted else "DENIED")
print("Final System State:", final_state)
print("Execution Log:")
for line in execution_log:
    print("  ", line)
print("Final Output:", f"Authentication {'succeeded' if access_granted else 'failed'} after {attempts_made} attempt(s); system is {final_state}")
```
